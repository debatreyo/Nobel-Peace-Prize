```python
import pandas as pd

import os
```

```python
data_winners = pd.read_csv(
    os.path.join(
        "..", "data",
        "Nobel Peace Winners.csv"
    )
)

data_nominees = pd.read_csv(
    os.path.join(
        "..", "data",
        "Nobel Peace Nominees.csv"
    )
)
```

```python
df_winners = data_winners.copy()
df_nominees = data_nominees.copy()
```

```python
df_winners.head()
```

```python
df_nominees.head()
```

```python
# create decade
df_nominees["Decade"] = (df_nominees["Year"] // 10) * 10
df_winners["Decade"] = (df_winners["Year"] // 10) * 10
```

```python
df_nominees
```

```python
# count nominees by decade × region
df_nominee_region_decade = (
    df_nominees.groupby(["Decade", "Region"])
    .size()
    .unstack(fill_value=0)
    .reset_index()
)

df_nominee_region_decade
```

```python
df_nominee_region_decade["Total"] = df_nominee_region_decade[[
    "Africa", "America", "Asia", "Europe", "Oceania"
]].sum(axis=1)

df_nominee_region_decade
```

```python
df_nominee_region_decade.to_csv(
    os.path.join(
            "..", "data",
            "Nobel Peace Nominees (Decade).csv"
            ),
            index=False
)
```

```python
# count winners by decade × region
df_winner_region_decade = (
    df_winners.groupby(["Decade", "Region"])
    .size()
    .unstack(fill_value=0)
    .reset_index()
)

df_winner_region_decade
```

```python
df_winner_region_decade["Total"] = df_winner_region_decade[[
    "Africa", "America", "Asia", "Europe"
]].sum(axis=1)

df_winner_region_decade
```

```python
df_winner_region_decade.to_csv(
    os.path.join(
            "..", "data",
            "Nobel Peace Winners (Decade).csv"
            ),
            index=False
)
```
